### 4. Análisis exploratorio inicial

Una vez validada la calidad y la integridad de la estructura de los datos, se procede a realizar el análisis exploratorio inicial. Este proceso consiste en examinar por separado las variables categóricas (calculando sus frecuencias absolutas y porcentuales) y las variables numéricas (obteniendo las principales medidas de tendencia central y dispersión: media, mediana, valores mínimos, máximos y desviación estándar). Esto permite comprender el comportamiento general de las calificaciones y los factores demográficos de los estudiantes.





```python
print("--- Frecuencias: Género (gender) ---")
print(df["gender"].value_counts())
print("\n--- Porcentajes: Género (gender) ---")
print(df["gender"].value_counts(normalize=True) * 100)

print("\n--- Frecuencias: Nivel educativo de los padres (parental level of education) ---")
print(df["parental level of education"].value_counts())


print("\n--- Estadísticas Descriptivas de las Variables Numéricas ---")
print(df[["math score", "reading score", "writing score"]].describe())
```

Análisis de variables categóricas (ejemplo: género y nivel educativo de los padres).

Análisis de variables numéricas (medidas estadísticas descriptivas).


**Pregunta:** ¿Qué hallazgos se observan en la distribución de las variables categóricas y numéricas?


A partir de la ejecución de las funciones estadísticas en Python, se desprenden los siguientes hallazgos analíticos sobre la muestra de 1,000 estudiantes:
* **Variables categóricas:** La distribución de género muestra un alto equilibrio, registrando 518 mujeres (51.8%) y 482 hombres (48.2%). En cuanto al nivel educativo de los padres, las categorías con mayor frecuencia son "some college" (226 casos) y "associate's degree" (222 casos), mientras que el nivel de "master's degree" presenta la menor representación (59 casos), evidenciando el trasfondo socioeducativo de los hogares.
* **Variables numéricas:** Las medidas descriptivas de los puntajes en matemáticas (math score), lectura (reading score) y escritura (writing score) revelan medias muy similares ubicadas entre 66 y 69 puntos, con una desviación estándar cercana a los 15 puntos. Se destaca un valor mínimo de 0 en matemáticas y un máximo de 100 en todas las materias, reflejando una amplia dispersión en los extremos del rendimiento estudiantil a pesar de la estabilidad de sus valores centrales (medianas de 66 en matemáticas, 70 en lectura y 69 en escritura).

### 5.Interpretación

En esta sección final se sintetizan los hallazgos del estudio de caso a partir de la interpretación analítica de las variables. El propósito es dar respuesta a las preguntas orientadoras del proceso, explicando el contexto del fenómeno, las variables clave, las relaciones detectadas entre ellas y su potencial aplicación en futuros modelos de Inteligencia Artificial y Machine Learning.

```python

correlacion = df[["math score", "reading score", "writing score"]].corr()
print("Matriz de Correlación de Puntajes:")
print(correlacion)
```

Análisis de correlación numérica entre los puntajes de los estudiantes.

Permite descubrir si los resultados de un examen están conectados con los de otro. Por ejemplo, ayuda a responder preguntas como: ¿Los estudiantes que sacan notas altas en lectura tienden a sacar notas altas también en escritura?

**Interpretación**: El resultado suele medirse con un coeficiente (como el coeficiente de correlación de Pearson) que va de -1 a 1:

Cercano a 1: Indica una correlación positiva fuerte (si una materia sube, la otra también sube de manera proporcional).

Cercano a 0: Indica que no hay relación aparente entre las variables.

Cercano a -1: Indica una correlación negativa (si una sube, la otra baja).
